# Data cleaning

#### Import library

```python
import pandas as pd
import pycountry
```

Normalization

```python
def to_iso3(name):
    # trasform country in alpha_3
    try:
        return pycountry.countries.lookup(name).alpha_3
    except:
        return None  # if the name is not found
custom_map = {
    "Turkey": "Türkiye",
    "UK": "United Kingdom",
    "USA": "United States",
    "USSR": "Russia",
    "Czech Republic": "Czechia",
    "South Korea": "Korea, Republic of",
    "North Korea": "Korea, Democratic People's Republic of",
    "Russia": "Russian Federation",
    "Yugoslavia": None,            # Historic town that no longer exists
    "Czechoslovakia": None         # Historic town that no longer exists
}
```

#### Read csv

```python
actors = pd.read_csv("../data/raw/static_data/actors.csv/actors.csv")
countries = pd.read_csv("../data/raw/static_data/countries.csv/countries.csv")
crew = pd.read_csv("../data/raw/static_data/crew.csv/crew.csv")
genres = pd.read_csv("../data/raw/static_data/genres.csv/genres.csv")
#languages = pd.read_csv("../data/raw/static_data/languages.csv/languages.csv")
#movies = pd.read_csv("../data/raw/static_data/movies.csv/movies.csv")
#posters = pd.read_csv("../data/raw/static_data/posters.csv/posters.csv")
#releases = pd.read_csv("../data/raw/static_data/releases.csv/releases.csv")
#studios = pd.read_csv("../data/raw/static_data/studios.csv/studios.csv")
#themes = pd.read_csv("../data/raw/static_data/themes.csv/themes.csv")
#the_oscar_awards = pd.read_csv("../data/raw/dynamic_data/the_oscar_awards/the_oscar_awards.csv")
#rotten_tomatoes_reviews = pd.read_csv("../data/raw/dynamic_data/rotten_tomatoes_reviews.csv")
```

### Study and cleaning csv

### Actors:

Print number duplicates, nan and empy

```python
print("Duplicates actors:", actors.duplicated().sum())
print("-------------------------\nNan actors:\n", actors.isnull().sum())
print("-------------------------\nColumn name rows with empty strings or only spaces:", (actors["name"].str.strip() == "").sum())
print("-------------------------\nColumn role rows with empty strings or only spaces:", (actors["role"].str.strip() == "").sum())
print("-------------------------\nCheck type columns")
actors.dtypes
```

### Countries:

Normalization country in iso_3

```python
#raplace country custom_map
countries["country"] = countries["country"].replace(custom_map)
#trasform country in iso 3
countries["country"] = countries["country"].str.strip().apply(to_iso3)
```

Print number duplicates, nan and empy

```python
print("Duplicates countries:", countries.duplicated().sum())
print("-------------------------\nNan countries:\n", countries.isnull().sum())
print("-------------------------\nColumn country rows with empty strings or only spaces:", (countries["country"].str.strip() == "").sum())
print("-------------------------\nCheck type columns")
countries.dtypes
```

Delete duplicates, because it would be a redundancy of data

```python
countries = countries.drop_duplicates()
countries = countries.dropna(subset=["country"])
```

Write new csv

```python
#countries.to_csv("../data/cleaned/countries_clean.csv", index=False)
```
